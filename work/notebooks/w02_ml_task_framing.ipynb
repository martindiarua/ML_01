## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule might say, "Refresh every page with declining traffic."

However, this ignores many other factors that influence whether refreshing a page is worthwhile.

For example, two pages may both have declining traffic, but one may receive thousands of monthly visits while the other receives only a few dozen. One page may also be several years old while another was recently published.

Machine learning can consider multiple signals simultaneously, including traffic trends, freshness, engagement, and other page characteristics, to estimate which pages are most valuable to refresh first.

Instead of replacing editors, the model helps them prioritise their work more consistently than a simple rule-based approach.
